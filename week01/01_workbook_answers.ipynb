### 문제 1
`print`로 네 이름과 좋아하는 동물을 **한 줄에 하나씩** 출력해 봐.

````python
print("수리")
print("고양이")
````

### 문제 2
결과를 **먼저 예상**해 봐. 그다음 아래 코드를 직접 쳐서 실행!
```python
print(9 / 3)
```

````python
print(9 / 3)
````

> **풀이** `3`이 아니라 `3.0`이 나와. 빗금 하나(`/`)로 나누면 딱 나누어떨어져도 항상 소수점이 붙어.

### 문제 3
사탕 23개를 4명이 똑같이 나눠 가지면, 한 명당 몇 개이고 몇 개가 남을까?

💡 힌트: 몫은 `//`, 나머지는 `%`

````python
print(23 // 4, 23 % 4)
````

> **풀이** 한 명당 5개, 남는 건 3개. 4 × 5 = 20, 23 − 20 = 3.

### 문제 4
2를 8번 곱한 수를 구해 봐.

💡 힌트: 거듭제곱은 `**`

````python
print(2 ** 8)
````

> **풀이** 2 ** 8 = 256. 2를 8번 곱한 수야.

### 문제 6
37은 짝수일까, 홀수일까? `%`로 확인해 봐.

💡 힌트: 2로 나눈 나머지를 보면 돼

````python
print(37 % 2)
````

> **풀이** 나머지가 1이니까 홀수. 나머지가 0이면 짝수야.

### 문제 7
영화가 135분짜리야. 몇 시간 몇 분일까?
