## 2. 사칙연산
따옴표 없이 식을 넣으면 파이썬이 계산해서 답을 보여 줘. 곱하기는 `*`, 나누기는 `/`.

```python
print(3 + 4)
print(7 - 2)
print(6 * 7)
print(7 / 2)
```

`/`로 나누면 항상 소수점이 붙어. 몫은 `//`, 나머지는 `%`, 거듭제곱은 `**`.

```python
print(7 // 2)
print(7 % 2)
print(2 ** 10)
```

### 계산 순서
수학처럼 곱하기·나누기를 더하기·빼기보다 먼저 해. 먼저 하고 싶은 건 괄호로 감싸.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
```

- 빛의 속도: 초속 약 29만 9792 km (코드로는 `299792`)

걸리는 시간(초) = 거리 ÷ 속도. 숫자에는 쉼표를 넣지 않아.

```python
print(149600000 // 299792)
```

499초를 분과 초로 바꾸려면 60으로 나눈 **몫**과 **나머지**를 쓰면 돼.

```python
print(499 // 60, "분", 499 % 60, "초")
```
