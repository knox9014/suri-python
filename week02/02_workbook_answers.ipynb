### 문제 5

### 문제 8
수업에서는 화씨 → 섭씨를 만들었지? 이번엔 거꾸로 **섭씨 → 화씨** 계산기를 만들어서 체온 36.5도를 넣어 봐. (화씨 = 섭씨 × 9 ÷ 5 + 32)

💡 힌트: 소수점을 입력할 거니까 `float`

````python
c = float(input("섭씨 온도는? "))
f = c * 9 / 5 + 32
print(round(f, 1), "도")
````

> **풀이** 36.5 × 9 ÷ 5 + 32 = 97.7. 수업에서 말한 '약 98도'가 맞았지?

### 문제 9
두 수를 입력받아서 **합**을 출력해 봐. 7과 5를 넣으면 12가 나와야 해. `int`를 빼면 어떻게 될지도 예상해 봐.

````python
a = int(input("첫 번째 수? "))
b = int(input("두 번째 수? "))
print(a + b)
````

> **풀이** `int`를 빼면 문자열끼리 이어 붙여서 `75`가 나와. 계산하려면 꼭 숫자로 바꿔야 해.

### 문제 10
초를 입력받아서 **몇 분 몇 초**인지 바꿔 봐. 1000을 넣으면 `16 분 40 초`.

💡 힌트: 1주차의 몫 `//`과 나머지 `%`

````python
s = int(input("몇 초? "))
print(s // 60, "분", s % 60, "초")
````

> **풀이** 60으로 나눈 몫이 분, 나머지가 초야.

### 문제 11
마라톤은 42.195 km야. km를 입력받아서 **마일**로 바꿔 봐. (1마일 = 약 1.609 km, 소수 첫째 자리까지)

💡 힌트: 마일 = km ÷ 1.609

````python
km = float(input("몇 km? "))
print(round(km / 1.609, 1), "마일")
````

> **풀이** 42.195 ÷ 1.609 = 약 26.2마일. 미국에서는 마라톤을 26.2마일이라고 불러.

## 수업 과제 정답
키를 cm로 입력받아서 피트와 인치로 바꾸기. 1인치 = 2.54 cm, 1피트 = 12인치.

````python
cm = float(input("키는 몇 cm? "))
inch = round(cm / 2.54)
print(inch // 12, "피트", inch % 12, "인치")
````

> **풀이** 170 ÷ 2.54 = 66.9인치, 반올림하면 67인치. 67을 12로 나눈 몫 5가 피트, 나머지 7이 인치야. `int()`로 소수점을 버리면 66인치가 돼서 1인치 틀려. 그래서 `round()`로 반올림!
