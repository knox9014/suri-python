## 1. 변수
`이름 = 값` 으로 쓰면 오른쪽 값에 왼쪽 **이름표**가 붙어. 이걸 **변수**라고 해.
- 여기서 `=`는 "같다"가 아니라 "이름표를 붙여라"라는 뜻이야.
- 이름을 부르면 그 값이 나와.

```python
distance = 149600000  # 태양~지구 거리(km)
speed = 299792  # 빛의 속도(km/초)
print(distance // speed)
```

이름표는 다른 값으로 옮겨 붙일 수 있어. 1편 과제였던 달까지의 거리로 바꿔 보자.

`speed`는 위 칸에서 붙였으니까 그대로 쓸 수 있어. 위 칸을 건너뛰면 `NameError`가 나니까, **위에서부터 차례대로** 실행해.

```python
distance = 384400  # 달~지구 거리(km)
print(distance / speed)
```

### 변수 이름 규칙
- 영어·숫자·밑줄(`_`)로 짓고, **숫자로 시작하면 안 돼** (`2nd` ✗)
- 대문자와 소문자는 다른 이름이야 (`Speed`와 `speed`는 달라)
- 무엇이 들었는지 알 수 있게 지어 (`d`보다 `distance`)

### 값 바꾸기
`count = count + 1`은 "지금 값에 1을 더하고, 그 결과에 이름표를 다시 붙여라"라는 뜻이야. 오른쪽부터 계산해.

```python
count = 0
count = count + 1
count = count + 1
print(count)
```

## 2. 자료형
값에는 종류가 있어. `type()`으로 확인할 수 있어.
- `int` 정수 (소수점 없는 수)
- `float` 실수 (소수점 있는 수)
- `str` 문자열 (따옴표로 감싼 글자, string의 줄임말)

```python
print(type(7))
print(type(3.5))
print(type("7"))
```

## 3. 입력 받기
`input()`을 실행하면 칸 아래에 입력 칸이 생겨. 값을 치고 **Enter**를 눌러. 괄호 안의 글은 질문으로 보여 줄 안내 문구야.

입력 칸이 떠 있는 동안에는 셀이 계속 실행 중이야. 입력을 끝내야 다음 칸을 실행할 수 있어.

```python
name = input("이름이 뭐야? ")
print("안녕,", name)
```

`input()`으로 받은 값은 숫자를 쳐도 **항상 문자열(`str`)**이야.

```python
temp = input("온도는? ")
print(type(temp))
print(temp + 1)
```

문자열을 숫자로 바꾸려면 `int()`(정수)나 `float()`(실수)로 감싸. 괄호가 두 겹이니까 닫는 괄호 개수를 잘 세어 봐.

```python
temp = int(input("온도는? "))
print(temp + 1)
```

## 4. 오늘의 질문 풀기: 화씨 → 섭씨
- 섭씨: 물이 0도에서 얼고 100도에서 끓어
- 화씨: 물이 32도에서 얼고 212도에서 끓어

**섭씨 = (화씨 − 32) × 5 ÷ 9**

`round(값, 1)`은 소수 첫째 자리까지 반올림해.

```python
fahrenheit = float(input("화씨 온도는? "))
celsius = (fahrenheit - 32) * 5 / 9
print(round(celsius, 1), "도")
```

## 5. 오류 하나 더
`int()`는 소수점이 있는 글자를 정수로 바꾸지 못해.

```python
print(int("3.5"))
```

소수점이 있으면 `float()`를 써.

```python
print(float("3.5"))
```
